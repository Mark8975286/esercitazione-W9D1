# Esercizio 1 · Variabili e tipi di dato (semplice · 20 minuti)

## Esercizio 1.1
**Domanda:** assegnare ciascun valore (numero di studenti di un corso, prezzo medio di un prodotto, nome di un cliente, se un ordine è stato consegnato, un valore assente) a una variabile con un nome descrittivo, rispettando le regole di naming.

```python
# Nomi descrittivi: iniziano con una lettera, solo caratteri alfanumerici e underscore (snake_case)
num_studenti = 25
prezzo_medio = 19.99
nome_cliente = "Laura Bianchi"
ordine_consegnato = True
valore_assente = None
```

## Esercizio 1.2
**Domanda:** verificare con `type()` il tipo di ciascuna variabile e stampare il risultato con `print()`.

```python
print("num_studenti      =", num_studenti, "->", type(num_studenti))
print("prezzo_medio      =", prezzo_medio, "->", type(prezzo_medio))
print("nome_cliente      =", nome_cliente, "->", type(nome_cliente))
print("ordine_consegnato =", ordine_consegnato, "->", type(ordine_consegnato))
print("valore_assente    =", valore_assente, "->", type(valore_assente))
```

## Esercizio 1.3
**Domanda:** indicare quale valore corrisponde a ciascuno dei cinque tipi di dato base: `int`, `float`, `str`, `bool`, `None`.

```python
corrispondenze = {
    "int": num_studenti,
    "float": prezzo_medio,
    "str": nome_cliente,
    "bool": ordine_consegnato,
    "None": valore_assente,
}

for tipo_atteso, valore in corrispondenze.items():
    print(f"{tipo_atteso:6} -> {valore!r:16} (type: {type(valore).__name__})")
```

| Tipo | Variabile | Valore |
|---|---|---|
| `int` | `num_studenti` | `25` |
| `float` | `prezzo_medio` | `19.99` |
| `str` | `nome_cliente` | `"Laura Bianchi"` |
| `bool` | `ordine_consegnato` | `True` |
| `None` (`NoneType`) | `valore_assente` | `None` |

### Definizione richiesta
**Variabile:** contenitore di dati identificato da un nome, che fa riferimento a un valore in memoria.

**Perché il tipo non è fissato alla dichiarazione:** Python ha una **tipizzazione dinamica**: il tipo appartiene al *valore*, non al nome della variabile, quindi lo stesso nome può contenere un `int` e subito dopo una `str`, come mostra la cella seguente.

```python
esempio = 42
print(esempio, type(esempio))
esempio = "quarantadue"
print(esempio, type(esempio))
```

# Esercizio 2 · Operatori e precedenza (semplice · 20 minuti)

## Esercizio 2.1
**Domanda:** calcolare il totale dell'ordine (`prezzo * quantita`) e poi applicare lo sconto, usando le parentesi per far calcolare prima il totale e poi la percentuale.

```python
prezzo = 45.90
quantita = 3
sconto_percentuale = 10
nome = "Mario"
cognome = "Rossi"

print(type(prezzo), type(quantita), type(sconto_percentuale))

totale = prezzo * quantita
totale_scontato = (prezzo * quantita) * (1 - sconto_percentuale / 100)

print("Totale ordine:", round(totale, 2))
print("Totale scontato:", round(totale_scontato, 2))
```

**Motivazione delle parentesi:** le parentesi raggruppano `prezzo * quantita` e `(1 - sconto_percentuale / 100)` in modo da calcolare prima il totale e il fattore di sconto, e solo dopo moltiplicarli, rendendo l'ordine dei calcoli esplicito invece di affidarlo alla precedenza degli operatori.

## Esercizio 2.2
**Domanda:** costruire la stringa di saluto concatenando nome e cognome con l'operatore `+`, inserendo uno spazio tra i due valori.

```python
print(type(nome), type(cognome))

saluto = "Ciao " + nome + " " + cognome
print(saluto)
```

## Esercizio 2.3
**Domanda:** usare un operatore di assegnazione composto (`+=`, `*=` o `-=`) per aggiornare la variabile `quantita`, simulando l'aggiunta di un prodotto all'ordine.

```python
print("Quantità prima:", quantita)
quantita += 1          # equivale a quantita = quantita + 1
print("Quantità dopo:", quantita)

totale_aggiornato = (prezzo * quantita) * (1 - sconto_percentuale / 100)
print("Nuovo totale scontato:", round(totale_aggiornato, 2))
```

### Definizione richiesta
**Operatore:** simbolo che esegue un'operazione su uno o più valori (operandi) e restituisce un risultato, ad esempio `+`, `*`, `==`, `+=`.

**Perché `//` e `%` danno risultati diversi da `/`:** `/` è la divisione vera e restituisce sempre un `float` con la parte decimale; `//` è la divisione intera e restituisce solo il quoziente arrotondato per difetto; `%` restituisce il resto della divisione intera.

```python
a, b = 7, 2
print("7 / 2  =", a / b, type(a / b))
print("7 // 2 =", a // b, type(a // b))
print("7 % 2  =", a % b, type(a % b))
```

# Esercizio 3 · Liste e tuple (semplice · 20 minuti)

## Esercizio 3.1
**Domanda:** sulla lista dei prezzi aggiungere un nuovo prezzo con `append()`, ordinarla con `sort()` e stampare il risultato.

```python
prezzi = [12.50, 8.00, 23.90, 5.50, 15.00]
coordinate_negozio = (45.0703, 7.6869)   # (latitudine, longitudine)

print(type(prezzi), type(coordinate_negozio))

prezzi.append(9.99)
prezzi.sort()          # modifica la lista in loco e restituisce None
print("Lista ordinata:", prezzi)
```

**Nota:** `print(prezzi.sort())` stamperebbe `None`, perché `sort()` modifica la lista originale senza restituirla; per questo l'ordinamento e la stampa sono su due righe separate.

## Esercizio 3.2
**Domanda:** recuperare il primo e l'ultimo prezzo della lista usando l'indexing, anche con indice negativo.

```python
primo_prezzo = prezzi[0]
ultimo_prezzo = prezzi[-1]
ultimo_con_len = prezzi[len(prezzi) - 1]

print("Primo prezzo (indice 0):", primo_prezzo)
print("Ultimo prezzo (indice -1):", ultimo_prezzo)
print("Ultimo prezzo (indice len-1):", ultimo_con_len)
```

## Esercizio 3.3
**Domanda:** sulla tupla delle coordinate provare a modificare il primo valore e riportare l'errore restituito da Python, spiegandone la causa.

```python
try:
    coordinate_negozio[0] = 45.1000
except TypeError as errore:
    print("Tipo di eccezione:", type(errore).__name__)
    print("Messaggio:", errore)

print("Tupla invariata:", coordinate_negozio)
```

**Causa dell'errore:** Python solleva un `TypeError` perché la tupla è **immutabile**: dopo la creazione i suoi elementi non possono essere sostituiti; per cambiare un valore bisogna creare una nuova tupla.

### Definizione richiesta
- **Lista:** sequenza **ordinata** e **mutabile** di elementi, scritta tra parentesi quadre `[ ]`.
- **Tupla:** sequenza **ordinata** e **immutabile** di elementi, scritta tra parentesi tonde `( )`.

Entrambe mantengono l'**ordine**; la proprietà che le distingue è la **modificabilità**.

# Esercizio 4 · Set e dizionari (avanzato · 30 minuti)

## Esercizio 4.1
**Domanda:** ottenere l'insieme dei voti unici a partire dalla lista, usando `set()`, e calcolare quanti valori duplicati sono stati eliminati.

```python
voti = [7, 8, 7, 9, 8, 10, 7]
voti_unici = set(voti)

print(type(voti), type(voti_unici))
print("Voti unici:", voti_unici)

duplicati_eliminati = len(voti) - len(voti_unici)
print("Elementi nella lista:", len(voti))
print("Elementi nel set:", len(voti_unici))
print("Duplicati eliminati:", duplicati_eliminati)
```

## Esercizio 4.2
**Domanda:** aggiungere alla rubrica un nuovo contatto e aggiornare il numero di un contatto già presente, usando l'assegnazione tramite chiave.

```python
rubrica = {"Anna": "1234", "Marco": "5678"}
print(type(rubrica))
print("Rubrica iniziale:", rubrica)

rubrica["Luca"] = "9012"   # chiave nuova -> aggiunge il contatto
rubrica["Anna"] = "4321"   # chiave esistente -> aggiorna il valore

print("Rubrica aggiornata:", rubrica)
```

## Esercizio 4.3
**Domanda:** usare `dict.get()` per cercare un contatto che non esiste in rubrica, confrontando il risultato con quello ottenuto usando `rubrica[chiave]`.

```python
contatto_cercato = "Giulia"

risultato_get = rubrica.get(contatto_cercato)
print("rubrica.get():", risultato_get)
print("rubrica.get() con default:", rubrica.get(contatto_cercato, "contatto non trovato"))

try:
    risultato_diretto = rubrica[contatto_cercato]
except KeyError as errore:
    print("rubrica[chiave] -> eccezione:", type(errore).__name__, errore)
```

**Differenza tra `get()` e accesso diretto:** `get()` restituisce `None` (o il valore di default indicato) quando la chiave non esiste, senza interrompere il programma; `rubrica["Giulia"]` invece solleva un `KeyError`, perché la chiave richiesta non è presente nel dizionario.

### Definizione richiesta
**Dizionario:** struttura dati **mutabile** che memorizza coppie **chiave: valore**, con chiavi uniche.

**Perché l'accesso avviene per chiave:** in un dizionario ogni valore è associato a un'etichetta (la chiave) e Python lo ritrova direttamente tramite quella; in una lista invece gli elementi sono identificati solo dalla loro posizione numerica (indice 0, 1, 2…).

# Esercizio 5 · Slicing e metodi combinati (avanzato · 30 minuti)

## Esercizio 5.1
**Domanda:** su `vendite` eseguire separatamente `vendite[::2]` e `vendite[::-1]`. Stampare entrambi i risultati.

```python
vendite = [120, 340, 275, 410, 90, 560, 205]
testo = "  Analisi Vendite Trimestre  "
print(type(vendite), type(testo))

passo_2 = vendite[::2]
passo_meno_1 = vendite[::-1]

print("vendite[::2]  =", passo_2)
print("vendite[::-1] =", passo_meno_1)
```

## Esercizio 5.2
**Domanda:** sulla stessa lista eseguire separatamente `vendite[:3]` e `vendite[4:]`. Stampare entrambi i risultati e il numero di elementi di ciascuno.

```python
prime_tre = vendite[:3]
dal_quinto = vendite[4:]

print("vendite[:3] =", prime_tre, "-> elementi:", len(prime_tre))
print("vendite[4:] =", dal_quinto, "-> elementi:", len(dal_quinto))
```

## Esercizio 5.3
**Domanda:** su `testo` applicare in sequenza `strip()`, poi `lower()` sul risultato ottenuto, poi `split()` sul risultato precedente. Stampare l'output dopo ciascuno dei tre passaggi.

```python
passo_strip = testo.strip()
print("1) strip():", repr(passo_strip))

passo_lower = passo_strip.lower()
print("2) lower():", repr(passo_lower))

passo_split = passo_lower.split()
print("3) split():", passo_split, type(passo_split))
```

### Definizione richiesta
**Slicing:** recupero di un sottoinsieme di elementi di una sequenza con la sintassi `[start:stop:step]`; restituisce sempre una nuova sequenza, senza modificare l'originale.

**Differenza tra step 2 e step -1:** con `step = 2` la sequenza viene letta da sinistra a destra prendendo un elemento ogni due (`[120, 275, 90, 205]`); con `step = -1` viene letta da destra a sinistra prendendo tutti gli elementi, quindi si ottiene la lista invertita (`[205, 560, 90, 410, 275, 340, 120]`).

# Esercizio 6 · Scelta della struttura dati (avanzato · 30 minuti)

## Esercizio 6.1
**Domanda:** l'elenco degli ordini ricevuti in un giorno, dove l'ordine di arrivo conta e più ordini possono avere lo stesso importo.

```python
ordini_giorno = [59.90, 120.00, 59.90, 35.50]
print(ordini_giorno, type(ordini_giorno))
ordini_giorno.append(80.00)      # arriva un nuovo ordine in coda
print("Dopo un nuovo ordine:", ordini_giorno)
```

**Struttura scelta: lista.** Proprietà determinante: **ordine** – la lista conserva la sequenza di arrivo e ammette valori duplicati (due ordini da 59.90).

## Esercizio 6.2
**Domanda:** le coordinate geografiche di un magazzino, che non devono cambiare dopo essere state registrate.

```python
coordinate_magazzino = (45.2210, 7.8250)
print(coordinate_magazzino, type(coordinate_magazzino))
```

**Struttura scelta: tupla.** Proprietà determinante: **modificabilità** – la tupla è immutabile, quindi le coordinate non possono essere alterate per errore.

## Esercizio 6.3
**Domanda:** l'elenco delle città già coperte da un servizio di consegna, dove serve verificare rapidamente se una città è già presente.

```python
citta_coperte = {"Torino", "Chieri", "Moncalieri", "Torino"}
print(citta_coperte, type(citta_coperte))
print("Chieri è coperta?", "Chieri" in citta_coperte)
print("Asti è coperta?", "Asti" in citta_coperte)
```

**Struttura scelta: set.** Proprietà determinante: **unicità** – il set non ammette duplicati (il secondo "Torino" viene scartato) e il controllo con `in` è molto veloce.

## Esercizio 6.4
**Domanda:** l'anagrafica di un fornitore, con nome, indirizzo e partita IVA associati a un'unica etichetta.

```python
fornitore = {
    "nome": "Logistica Nord S.r.l.",
    "indirizzo": "Via Roma 10, Torino",
    "partita_iva": "01234567890",
}
print(type(fornitore))
for chiave, valore in fornitore.items():
    print(f"{chiave}: {valore}")
```

**Struttura scelta: dizionario.** Proprietà determinante: **unicità delle chiavi** – ogni dato è associato a un'etichetta univoca (`nome`, `indirizzo`, `partita_iva`) e si recupera per chiave, non per posizione.

### Riepilogo
| Scenario | Struttura | Proprietà determinante |

| 1. Ordini del giorno | Lista | Ordine (con duplicati ammessi) |
| 2. Coordinate magazzino | Tupla | Modificabilità (immutabile) |
| 3. Città coperte | Set | Unicità |
| 4. Anagrafica fornitore | Dizionario | Unicità delle chiavi (accesso per etichetta) |
